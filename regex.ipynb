## understanding regex patterns 

a `regex pattern` is made up of characters and symbols that decribe what we want to find

there are two important thing to understand :
1. literal character math themselves
2. special regex symbols have meaning

### literal characters 
`literal characters ` are ordinry characters that match exactly what they represent

```python
text = "i am learning python"
result = re.search(r"python", text)
print(result.group())
```

## metacharacters
these are symbols that have a special meaning in regex

they allows us search patterns instead of exact character

## the dot(.)

```python
text = "cat, cut, bat, mat"
result = re.findall(r"[c.t]", text)
```

### character class

`character class` allow us to specify whicch character should match

they are written using square brackets[]

```python
text = "cat, cut, cot, caat, bat, mat"
result = re.findall(r"c[ao]t",text)
print(result)
```

## character ranges


```python
text = "mary123"
letters = re.findall(r"[a-zA-Z]", text)
digits = re.findall(r"[0-9]",text)
print(digits)
print(letters)
```
