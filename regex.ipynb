## regular expression 

`regular expression` are patterns used to search match extract validate and manipulate text.

Instead of searching for one exact value, Regex allows us to describe a pattern that different text values can follow.

## Why Do We Need Regex?
Python provides string methods such as find(),  replace(),  split(),  and the in operator.

These are useful when we know exactly what we are looking for.

```python
import re
```

```python
text = "i am learning python"
print("python" in text)
```

```python
text = "mary scored 85, john scored 90, and ann scored 78"
```

```python
scores = re.findall(r"\d+", text)
print(scores)
```

We want to extract all the scores without searching for 85, 90, or 78 individually.

Instead, we can describe what a score looks like: a sequence of digits.

## metacharacters

so far, we have searched for exact words.

But what if we want to find text that follows a particular structure rather than an exact word?

Regex provides special symbols called metacharacters.

Metacharacters allow us to describe more flexible patterns.

Metacharacters are special characters in Regular Expressions (Regex) that have a special meaning. They help us create patterns to search for specific characters, words or text.

these are symbols that have a special meaning in regex

they allows us search patterns instead of exact character

Common Regex Metacharacters 

| Metacharacter | Meaning | Example | What it matches | | ------------- | ------------------------------------------------------------- | ----------- | ------------------------- | | . | Matches any single character except a newline. | c.t | cat, cut, cot | | [] | Matches any one character inside the brackets. | [abc] | a, b, or c | | [^...] | Matches any character not listed inside the brackets. | [^abc] | d, e, f, etc. | | () | Groups parts of a pattern together. | (cat) | cat | | \\| | Matches either pattern (OR). | cat\\|dog | cat or dog | | \ | Escapes a special character or introduces a special sequence. | \. | A literal full stop (.) |



Note: To match a metacharacter as an ordinary character, use a backslash. For example, \. matches an actual full stop.

## Regex (Regular Expression) Metacharacters

| Symbol / Pattern | Name | Meaning | Example | Matches |
|---|---|---|---|---|
| `.` | Dot | Any character except newline | `a.c` | abc, a1c, a-c |
| `^` | Start anchor | Matches the beginning of a string | `^Hello` | Hello world |
| `$` | End anchor | Matches the end of a string | `world$` | Hello world |
| `*` | Star | 0 or more repetitions | `ab*` | a, ab, abb |
| `+` | Plus | 1 or more repetitions | `ab+` | ab, abb |
| `?` | Question mark | 0 or 1 repetition (optional) | `ab?` | a, ab |
| `{n}` | Exact repetition | Exactly n repetitions | `a{3}` | aaa |
| `{n,m}` | Repetition range | Between n and m repetitions | `a{2,4}` | aa, aaa, aaaa |
| `{n,}` | Minimum repetition | n or more repetitions | `a{2,}` | aa, aaa, aaaa |
| `[]` | Character set | Matches any character inside | `[abc]` | a, b, c |
| `[^]` | Negation | Matches characters NOT in set | `[^abc]` | d, x, 1 |
| `[-]` | Range inside set | Matches a range of characters | `[a-z]` | a to z |
| `()` | Group | Groups patterns together | `(abc)` | abc |
| `|` | Alternation (OR) | Matches either pattern | `cat|dog` | cat or dog |
| `\` | Escape character | Makes special symbols literal | `\.` | A literal dot |
| `\d` | Digit | Any digit (0–9) | `\d` | 5 |
| `\D` | Non-digit | Any character except digits | `\D` | A, @ |
| `\w` | Word character | Letters, digits, underscore | `\w` | a, 7, _ |
| `\W` | Non-word character | Anything except word characters | `\W` | @, !, space |
| `\s` | Whitespace | Space, tab, newline | `\s` | space |
| `\S` | Non-whitespace | Any character except whitespace | `\S` | A, 5, @ |
| `\b` | Word boundary | Boundary between word and non-word | `\bcat\b` | cat (whole word) |
| `\B` | Non-word boundary | Position that is NOT a word boundary | `\Bcat` | part of a larger word |
| `\A` | String start | Matches the absolute beginning | `\AHello` | Hello |
| `\Z` | String end | Matches the absolute end | `world\Z` | world |
| `(?i)` | Ignore case | Case-insensitive matching | `(?i)hello` | Hello, HELLO |
| `(?:...)` | Non-capturing group | Groups without saving a capture | `(?:ab)+` | ab, abab |
| `(?=...)` | Positive lookahead | Checks what comes next without consuming it | `a(?=b)` | a in ab |
| `(?!...)` | Negative lookahead | Checks that what comes next is NOT present | `a(?!b)` | a in ac |
| `(?<=...)` | Positive lookbehind | Checks what comes before | `(?<=\$)\d+` | 25 in $25 |
| `(?<!...)` | Negative lookbehind | Checks that what comes before is NOT present | `(?<!\$)\d+` | 25 not preceded by $ |

### character class

So far, we have used the dot to match any single character.

But sometimes we want to be more specific.

For example, suppose we want to match a word that starts with `c` and ends with `t`, but the middle character must be either `a` or `o`.

We can use `character classes`.

`character class` allow us to specify whicch character should match

they are written using square brackets[]

```python
text = "cat, cut, cot, caat, bat, mat"
result = re.findall(r"c[ao]t",text)
print(result)
```

let's break down the pattern:

- `c` means the text must begin with c.
- `[ao]` means the next character can be either a or o.
- `t` means the text must end with t.

Square brackets match one character from the options provided.

Unlike the dot, which matches any single character, `[ao]` only matches `a` or `o`.

## character ranges
instead of listing every character individually, we can specify a range.

| Character Type | Start | End | Range in Python |
|---|---:|---:|---|
| Uppercase letters (A-Z) | 65 | 90 | `range(65, 91)` |
| Lowercase letters (a-z) | 97 | 122 | `range(97, 123)` |
| Digits (0-9) | 48 | 57 | `range(48, 58)` |
| Space | 32 | 32 | `range(32, 33)` |
| Control characters | 0 | 31 | `range(0, 32)` |
| Punctuation and symbols | 33 | 126 | Various ranges |

```python
text = "mary123"
letters = re.findall(r"[a-zA-Z]", text)
digits = re.findall(r"[0-9]",text)
print(digits)
print(letters)
```

The pattern `[a-zA-Z]` matches one letter at a time, whether uppercase or lowercase.

The pattern `[0-9]` matches one digit at a time.
