### Python Data Structures: Lists, Tuples, Sets, and Dictionaries

##Introduction to Data Structures

Imagine you are an instructor responsible for keeping track of 100 students. You need to store each student's name.

You could create a separate variable for every student:

student1 = "Mary"

student2 = "Brian"

student3 = "John"

student4 = "Faith"

student5 = "Peter"
If we have 100 students, how many variables would we need? What if 20 more students joined tomorrow?

Challenges if we have 100 students
We would have keep creating new variables
Displaying every student's name would require referencing every variable separately
Counting, searching, updating, or removing students would become inconvenient
The code would become difficult to manage
The Solution: An Organised Class Register
Instead of creating a separate variable for every student, we can store all the names together in one list:

students = ["Mary", "Brian", "John", "Faith", "Peter"]
What has changed? We now have one variable, students, holding a collection of names. We can manage the collection together.

```python
students = ["Mary", "Brian", "John", "Faith", "Peter"]

# Add a new student
students.append("Jane")

# Find the number of students
print(len(students))

# Display all the students
for student in students:
    print(student)
```

## Lists

A list is an ordered and changeable collection of elements

Lists allow duplicate values and can store different data types

## Creating a List

Lists are created using square brackets []

```python
students = ["Mary", "Brian", "John", "Faith", "Peter"]

scores = [85, 90, 75, 85]

mixed = ["Mary", 22, True, 65.6]
```

### Characteristics of Lists

- Ordered: Elements maintain their insertion order.

- Mutable: Elements can be added, removed, or changed.

- Allow duplicates: The same value can appear more than once.

- Indexed: Elements can be accessed using their positions.

- Allow different data types: A list can contain strings, numbers, booleans, and other objects.

### Common List Methods

# Common Python List Methods

| **Method** | **Purpose** | **Example Usage** |
|---|---|---|
| `append()` | Adds an item to the end of a list. | `numbers.append(10)` |
| `extend()` | Adds multiple items to the end of a list. | `numbers.extend([20, 30])` |
| `insert()` | Adds an item at a specific position. | `numbers.insert(1, 15)` |
| `remove()` | Removes the first occurrence of a specified value. | `numbers.remove(10)` |
| `pop()` | Removes and returns an item at a specified index. | `numbers.pop(1)` |
| `clear()` | Removes all items from a list. | `numbers.clear()` |
| `index()` | Returns the index of the first occurrence of a value. | `numbers.index(20)` |
| `count()` | Counts the number of occurrences of a value. | `numbers.count(10)` |
| `sort()` | Sorts the list in ascending order by default. | `numbers.sort()` |
| `reverse()` | Reverses the order of items in a list. | `numbers.reverse()` |
| `copy()` | Creates a copy of a list. | `new_numbers = numbers.copy()` |




Note: Most methods that modify a list (append(),  insert(),  extend(),  remove(),  sort(),  reverse(),  and  clear()) modify the original list and return None. pop() returns the removed element,  while count()  and index() return values without modifying the list.

```python
numbers = [10, 20, 30, 40, 50]
numbers.append(60)
numbers.insert(6, 70)
numbers.extend([80, 90])
numbers.remove(90)
numbers.pop()
numbers.reverse()
numbers.sort()

numbers
```

```python
numbers = [70, 60, 50, 40, 30, 20, 10]
# numbers_reveresed = numbers.reverse()
# print(numbers_reveresed)
numbers_reveresed = numbers[::-1]
print(numbers_reveresed)
```

### Tuples

A tuple is an ordered collection of elements that cannot be changed after creation .

Tuples are created using parentheses ().

```python
student = ("Asma", 22, "Data Science", 85)
student
```

### Characteristics of Tuples

- Ordered: Elements maintain their insertion order.

- Immutable: Elements cannot be changed, added, or removed after creation.

- Allow duplicates: The same value can appear more than once.

- Indexed: Elements can be accessed using their positions.

- Allow different data types: A tuple can store different types of values.

### Accessing Tuple Elements

Tuples support indexing and slicing just like lists.

```python
student = ("Mary", 22, "Data Science", 85)

print(student[0])
print(student[2])
print(student[-1])
```

```python
### Tuple Immutability

scores = (70, 80, 90)
scores[0] = 100
```

```python
# 7.Create a tuple to store mathematical operators (+, -, *, /) and write a program that allows users to choose an operator and perform basic calculations with two numbers.
# Create a tuple of mathematical operators
operators = ("+", "-", "*", "/")

# Ask the user to enter two numbers
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Display available operators
print("Available operators:", operators)

# Ask the user to choose an operator
operator = input("Choose an operator (+, -, *, /): ")

# Perform the calculation
if operator == "+":
    result = num1 + num2
    print("Result:", result)

elif operator == "-":
    result = num1 - num2
    print("Result:", result)

elif operator == "*":
    result = num1 * num2
    print("Result:", result)

elif operator == "/":
    if num2 != 0:
        result = num1 / num2
        print("Result:", result)
    else:
        print("Error: Cannot divide by zero!")

else:
    print("Invalid operator!")
```

### set

A set is a collection of unique elements .

Sets are created using curly brackets {} or set constructor set()

### Characteristics of Sets

- Unordered: Sets do not guarantee element order.

- Mutable: Elements can be added or removed.

- No duplicates: Every element must be unique.

- Not indexed: Elements cannot be accessed using positions.

- Elements must be hashable: Mutable collections such as lists and

### Other Useful Set Methods

# Python Set Methods

| Method | Meaning | Example |
|---|---|---|
| `add()` | Adds one item to a set. | `my_set.add(5)` |
| `update()` | Adds multiple items to a set. | `my_set.update([6, 7, 8])` |
| `remove()` | Removes an item. Gives an error if it doesn't exist. | `my_set.remove(3)` |
| `discard()` | Removes an item without an error if it doesn't exist. | `my_set.discard(10)` |
| `pop()` | Removes and returns an arbitrary item. | `my_set.pop()` |
| `clear()` | Removes all items from a set. | `my_set.clear()` |
| `copy()` | Creates a copy of a set. | `new_set = my_set.copy()` |
| `union()` | Combines two sets without duplicates. | `A.union(B)` |
| `intersection()` | Returns items found in both sets. | `A.intersection(B)` |
| `difference()` | Returns items in the first set but not the second. | `A.difference(B)` |
| `symmetric_difference()` | Returns items that are in either set, but not both. | `A.symmetric_difference(B)` |
| `intersection_update()` | Keeps only common items and updates the original set. | `A.intersection_update(B)` |
| `difference_update()` | Removes items found in another set. | `A.difference_update(B)` |
| `symmetric_difference_update()` | Keeps items that are not common and updates the original set. | `A.symmetric_difference_update(B)` |
| `issubset()` | Checks if all items in one set are in another. | `A.issubset(B)` |
| `issuperset()` | Checks if a set contains all items of another set. | `A.issuperset(B)` |
| `isdisjoint()` | Checks if two sets have no common items. | `A.isdisjoint(B)` |

## Important Things to Remember

1. Sets do not allow duplicate values.
2. Sets are unordered.
3. Sets are created using `{}` or `set()`.
4. Sets are mutable, meaning you can add or remove items.
5. Set items must be immutable (e.g. strings, numbers, tuples).

### Set Operations

```python
python_students = {"Asma", "Jane", "Mark", "Wayne"}
sql_students = {"Jennifer", "Nasra", "Sahal", "Ceci", "Wayne"}

# Union - Returns all the unique elements from both sets
# print(python_students.union(sql_students))

# Intersection - Returns elements found in both sets 
print(python_students.intersection(sql_students))

# Difference - Returns the elements found in the first set but not in the second
print(python_students.difference(sql_students))

# Symmetric Difference - Returns elements found in either set, but not in both
print(python_students.symmetric_difference(sql_students))
```
