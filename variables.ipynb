### Making Random Results Reproducible

Randomness can make experiments difficult to reproduce

Imagine you are developing a Python program that randomly generates exam scores for 5 students.

You want to test how your program works, but every time you run it, the students receive different scores. This makes it difficult to compare results, identify errors, or investigate why a particular student received a certain score.

We can use random.seed() to make random results reproducible 

````python
random.seed(5)
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(numbers)
````

### Lambda Functions

A lambda function is a small anonymus function defined using the lambda keyword

Unlike regular functions, lambda functions are written in a single expression and do not require a function name in their definition

## Syntax

lambda arguments: expression


## example: squaring number

````python
# Using a regular function
def square(number):
    return number ** 2

result = square(4)
print("Result:", result)
````

````python
# Using a lambda function

square = lambda number: number ** 2

print(square(4))
````

````python
# Using lambda with multiple arguments

add = lambda a, b: a + b 

add(10, 20)
````

### Lambda With map()

The map() function applies a function to every item in an iterable

````python
numbers = [1, 2, 3, 4, 5]

square = list(map(lambda digit: digit ** 2, numbers))

print(square)
````

### Lambda With filter()

````python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
````

### Why Should We Use Lambda Functions?

+ For short operations

+ When passing a simple function to another function

+ When sorting data using a particular column

+ When using functions such as map() and filter()
