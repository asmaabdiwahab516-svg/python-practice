## why do we need file handling

file handling allows python program to:

- read information that has already been saved

- create new file 

- write information into the file 

- add new information without deleting existing content


### opening fie 
python uses the `open()` function to open a file so that we can read and write information

**syntax**

```python``
file = open("filename.txt","mode")

the `mode` determines how we interact with file 

### common file mode

# Common File Modes in Python

| File Mode | Name | Meaning |
|-----------|------|---------|
| "r" | Read | Opens a file for reading |
| "w" | Write | Writes content and replaces old content |
| "a" | Append | Adds new content at the end of a file |
| "x" | Create | Creates a new file |
| "r+" | Read and Write | Allows reading and writing |
| "w+" | Write and Read | Allows reading and writing but deletes old content |
| "a+" | Append and Read | Allows reading and adding content |

## Easy Way to Remember

| Mode | Remember |
|------|----------|
| r | Read |
| w | Write |
| a | Add |
| x | Create |


`nb` be caraful when using `w` because it replace the existing content

### reading from a file 

we use `r` mode to retrieve information from an existing file

````python
with open("student.txt", "r")as file:
    content = file.read()
    print(content)
````
