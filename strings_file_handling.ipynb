### string travesal

`travesal` means visiting each character in a string one at a time

```python
course = "python"

for i in course :
    #print(i)
    #print(len(course))
    print(i.count("a"))
```

```python
response = "data analysis is about finding pattern"
count = 0
for character in response:
    if character == "a":
        count += 1
    
```
